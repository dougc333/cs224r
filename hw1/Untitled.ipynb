```python
import gymnasium as gym
import numpy as np
import imageio.v2 as imageio

def replay_actions_to_video(path, out_mp4="replay.mp4", env_id="Ant-v4", fps=30):
    env = gym.make(env_id, render_mode="rgb_array")
    obs, info = env.reset()

    actions = np.asarray(path["action"])
    frames = []

    for a in actions:
        frame = env.render()
        if frame is not None:
            frames.append(frame)
        obs, r, terminated, truncated, info = env.step(a)
        if terminated or truncated:
            break

    env.close()

    with imageio.get_writer(out_mp4, fps=fps, codec="libx264") as w:
        for f in frames:
            w.append_data(np.asarray(f, dtype=np.uint8))

    print("Wrote", out_mp4)

#what is the difference? hard to tell maybe one has dead leg syndrome more than the other
#replay_actions_to_video(obj[0], "replay0.mp4", env_id="Ant-v4")
replay_actions_to_video(obj[1], "replay1.mp4", env_id="Ant-v4")
```
