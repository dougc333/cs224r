```python


import os
import gymnasium as gym
import torch
import time
import imageio
import numpy as np
import json
from collections import deque
from stable_baselines3 import PPO
from stable_baselines3.ppo import MlpPolicy
from stable_baselines3.common.vec_env import SubprocVecEnv, VecMonitor, VecNormalize
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.callbacks import BaseCallback
from stable_baselines3.common.logger import Video


startTime = time.time()

LOG_DIR='/content/hopper'
VIDEO_DIR = f"{LOG_DIR}/videos"
ENV="Hopper-v5"
NENV_TRAIN = 16
NENV_EVAL= 4
os.makedirs(LOG_DIR, exist_ok=True)
os.makedirs(VIDEO_DIR, exist_ok=True)

#for colab
os.environ["MUJOCO_GL"] = "egl"          # use EGL (headless GPU)
#for colab
os.environ["PYOPENGL_PLATFORM"] = "egl"

startTime = time.time()


#2 separate env
#train_env:
#    stochastic
#    normalization observations, normalized_obs = (obs - running_mean) / running_std
# want to center at origin w unit variance for GD. VecNormalize provides normalization. If
#we use dummyvec and dont call vecnorm then there is no obs normalization.
#    parallel rollouts

#eval_env:
#    deterministic
#    normalization frozen.
#    video generation
#    checkpoint and scoring


def make_train_env(env_id, n_envs):

    env = make_vec_env(
        env_id,
        n_envs=n_envs,
        vec_env_cls=SubprocVecEnv,   # ⭐ parallel rollouts
    )

    env = VecNormalize(
        env,
        norm_obs=True,              # ⭐ normalize observations
        norm_reward=False,          # ⭐ typical for locomotion PPO
        clip_obs=10.0,
    )
    env = VecMonitor(env)
    return env

train_env =  make_train_env(env_id=ENV, n_envs=NENV_TRAIN)


def make_eval_env(env_id, train_env, n_envs=NENV_EVAL):

    eval_env = make_vec_env(
        env_id,
        n_envs=n_envs,
        vec_env_cls=SubprocVecEnv,
        env_kwargs={"render_mode": "rgb_array"},
    )

    eval_env = VecNormalize(
        eval_env,
        norm_obs=True,
        norm_reward=False,
    )
    eval_env = VecMonitor(eval_env)

    # ⭐ CRITICAL
    eval_env.obs_rms = train_env.obs_rms
    eval_env.training = False
    eval_env.norm_reward = False

    return eval_env

eval_env =  make_eval_env(env_id=ENV, train_env=train_env, n_envs=4)

policy_kwargs = dict(
    ortho_init = True,
    net_arch=dict(
        activation_fn=torch.nn.Tanh,
        pi=[256, 256],
        vf=[256, 256]
    )
)
model = PPO(
    policy=MlpPolicy,
    env=train_env,
    n_steps=2048,
    batch_size = 128,
    n_epochs=5,
    learning_rate=2e-4,
    gamma=0.99,
    gae_lambda=0.95,
    clip_range=0.2,
    target_kl=0.01,
    ent_coef=0.01,
    verbose=1,
    vf_coef=0.5,
    max_grad_norm=0.5,
    policy_kwargs=policy_kwargs,
    device="cuda" if torch.cuda.is_available() else "cpu",
    tensorboard_log=LOG_DIR,
)
#Print the full structure of the neural network
print(f"model.policy:{model.policy}")


class CheckpointAndMetadataCallback(BaseCallback):
    def __init__(self, save_freq=500_000, save_dir="checkpoint_dir", verbose=1):
        super().__init__(verbose)
        self.save_freq = int(save_freq)
        self.save_dir = save_dir
        os.makedirs(self.save_dir, exist_ok=True)
        self.tb_dir = None
        self.ckpt_dir = None

    def _on_training_start(self) -> None:
        # Now SB3 has set self.model and self.logger
        self.tb_dir = self.logger.dir  # or: self.model.logger.dir
        print(f"tb_dir: {self.tb_dir}")

        self.ckpt_dir = os.path.join(self.tb_dir, "checkpoints")
        os.makedirs(self.ckpt_dir, exist_ok=True)
        print(f"created ckpt_dir: {self.ckpt_dir}")
        if self.verbose:
            print(f"Checkpoint dir: {self.ckpt_dir}")

    @staticmethod
    def _find_vecnormalize(env):
        """Walk VecEnvWrapper chain to find VecNormalize, or return None."""
        cur = env
        # SB3 VecEnv wrappers typically expose `.venv`
        while hasattr(cur, "venv"):
            if isinstance(cur, VecNormalize):
                return cur
            cur = cur.venv
        # Also check final object
        if isinstance(cur, VecNormalize):
            return cur
        return None

    def _on_step(self) -> bool:
        if self.num_timesteps >= 0 and (self.num_timesteps % self.save_freq) == 0:
            t = int(self.num_timesteps)
            model_path = os.path.join(self.ckpt_dir, f"model_t{t}")
            # Save model
            path = os.path.join(self.ckpt_dir, f"model_step_{self.num_timesteps}")
            self.model.save(path)
            print(f"Saved checkpoint: {path}.zip")
            if self.verbose:
                print(f"Saved model → {model_path}.zip")

            # Save VecNormalize stats (CRITICAL if you use VecNormalize)
            env = self.model.get_env()
            if env is not None:
                vecnorm = CheckpointAndMetadataCallback._find_vecnormalize(env)
                if vecnorm is not None:
                    vn_path = os.path.join(self.ckpt_dir, f"vecnormalize_t{t}.pkl")
                    vecnorm.save(vn_path)
                    if self.verbose:
                        print(f"Saved VecNormalize → {vn_path}")


            # Save simple JSON metadata
            progress = getattr(self.model, "_current_progress_remaining", 1.0)

            meta = {
            "timesteps": int(t),
            "policy": model.policy.__class__.__name__,
            "n_steps": int(self.model.n_steps),
            "batch_size": int(self.model.batch_size),
            "n_epochs": int(self.model.n_epochs),
            "gamma":float(self.model.gamma),
            "gae_lambda":float(self.model.gae_lambda),
            "learning_rate": float(self.model.lr_schedule(progress)),
            "clip_range": float(
              self.model.clip_range(progress)
              if callable(self.model.clip_range)
              else self.model.clip_range
            ),
            "ent_coef": float(self.model.ent_coef),
            "target_kl": (
              float(self.model.target_kl)
              if self.model.target_kl is not None
              else None
              ),
            "vf_coef":float(self.model.vf_coef),
            "max_grad_norm":float(self.model.max_grad_norm),
            "n_env_training": int(NENV_TRAIN),
            "n_env_eval": int(NENV_EVAL),
            }

            # ⭐ Save per-checkpoint metadata
            json_path = os.path.join(self.ckpt_dir, f"config_t{t}.json")

            with open(json_path, "w") as f:
              json.dump(meta, f, indent=2)

            if self.verbose:
              print(f"[ckpt] saved metadata → {json_path}")

        return True

    def _on_training_end(self) -> None:
        # final checkpoint
        t = int(self.num_timesteps)
        model_path = os.path.join(self.save_dir, f"model_final_t{t}.zip")
        self.model.save(model_path)
        env = self.model.get_env()
        if isinstance(env, VecNormalize):
            env.save(os.path.join(self.save_dir, f"vecnormalize_final_t{t}.pkl"))
        if self.verbose:
            print(f"[ckpt] saved final {model_path}")

class LogHopperJointsCallback(BaseCallback):
    """
    Hopper-v4/v5 action space is typically 3-dim:
      0: thigh joint motor
      1: leg joint motor
      2: foot joint motor

    Logs mean |action| per rollout and (optionally) per step.
    """
    def __init__(self, log_per_step: bool = False, verbose: int = 0):
        super().__init__(verbose)
        self.log_per_step = log_per_step
        self.joint_names = ["thigh", "leg", "foot"]

    def _on_step(self) -> bool:
        if not self.log_per_step:
            return True

        actions = self.locals.get("actions")  # shape (n_envs, 3) for on-policy
        if actions is None:
            return True

        mean_abs = np.mean(np.abs(actions), axis=0)  # (3,)
        for i, name in enumerate(self.joint_names):
            self.logger.record(f"joints_step/mean_abs_action/{name}", float(mean_abs[i]))
        self.logger.record("custom_step/total_activity", float(np.mean(mean_abs)))
        return True

    def _on_rollout_end(self) -> None:
        # rollout buffer actions: (n_steps, n_envs, act_dim)
        acts = self.model.rollout_buffer.actions
        acts = acts.reshape(-1, acts.shape[-1])  # (n_steps*n_envs, 3)

        mean_abs = np.mean(np.abs(acts), axis=0)
        std = np.std(acts, axis=0)

        self.logger.record("custom/total_activity", float(np.mean(mean_abs)))

        for i, name in enumerate(self.joint_names):
            self.logger.record(f"joints/mean_abs_action/{name}", float(mean_abs[i]))
            self.logger.record(f"joints/std_action/{name}", float(std[i]))
            self.logger.record(f"joints/mean_action/{name}", float(np.mean(acts[:, i])))
            self.logger.record(f"joints/max_abs_action/{name}", float(np.max(np.abs(acts[:, i]))))


class PPOMetricsCallback(BaseCallback):
    """
    Logs PPO + rollout + episode diagnostics to SB3 logger (TensorBoard compatible).

    Requires VecMonitor/Monitor so info["episode"] exists.

    Metrics (examples):
      - custom/ep_len_mean_window. #average episode length
      - custom/ep_rew_mean_window. # average episode reward

      - custom/ep_len_frac_horizon
      - custom/frac_time_limit
      - custom/frac_terminated

      - debug/actions/mean_abs
      - debug/actions/std
      - debug/actions/max_abs
      - debug/actions/dim_i_mean_abs, std, max_abs (per action dimension)

      - debug/policy/entropy_mean (estimated from rollout obs)
      - debug/policy/approx_kl (if available from SB3 train logs; else computed approx optional)

      - debug/value/vpred_mean, vpred_std
      - debug/adv/mean, adv/std (from rollout buffer)

      - debug/obs/mean, obs/std, obs/clip_frac (after normalization if VecNormalize)
      - debug/numerics/has_nan_obs, has_nan_actions

    Optional: write JSONL line per rollout_end for easy React tailing.
    """
    def __init__(
        self,
        horizon: int = 1000,
        window: int = 200,
        entropy_batch: int = 2048,
        log_jsonl_path: str | None = None,
        log_action_dims: bool = True,
        verbose: int = 0,
    ):
        super().__init__(verbose)
        self.horizon = horizon
        self.window = window
        self.entropy_batch = entropy_batch
        self.log_jsonl_path = log_jsonl_path
        self.log_action_dims = log_action_dims

        self.ep_lens = deque(maxlen=window)
        self.ep_rews = deque(maxlen=window)
        self.was_time_limit = deque(maxlen=window)
        self.was_terminated = deque(maxlen=window)

        self._jsonl_f = None

    def _init_callback(self) -> None:
        if self.log_jsonl_path is not None:
            os.makedirs(os.path.dirname(self.log_jsonl_path), exist_ok=True)
            self._jsonl_f = open(self.log_jsonl_path, "a", buffering=1)

    def _on_step(self) -> bool:
        infos = self.locals.get("infos", [])
        for info in infos:
            ep = info.get("episode", None)
            if ep is None:
                continue

            l = float(ep.get("l", 0.0))
            r = float(ep.get("r", 0.0))
            self.ep_lens.append(l)
            self.ep_rews.append(r)

            truncated = bool(info.get("TimeLimit.truncated", False))
            self.was_time_limit.append(truncated)
            self.was_terminated.append(not truncated)

        # Light per-step checks (cheap)
        actions = self.locals.get("actions", None)
        if actions is not None:
            self.logger.record("debug/numerics/has_nan_actions", float(np.isnan(actions).any()))
        obs = self.locals.get("new_obs", None)
        if obs is not None:
            try:
                self.logger.record("debug/numerics/has_nan_obs", float(np.isnan(obs).any()))
            except Exception:
                pass

        return True

    def _on_rollout_end(self) -> None:
        rb = self.model.rollout_buffer

        # -------------------------
        # Episode-window metrics
        # -------------------------
        if len(self.ep_lens) > 0:
            lens = np.asarray(self.ep_lens, dtype=np.float32)
            rews = np.asarray(self.ep_rews, dtype=np.float32)

            frac_h = float(np.mean(lens >= self.horizon))
            frac_tl = float(np.mean(np.asarray(self.was_time_limit, dtype=np.float32))) if len(self.was_time_limit) else 0.0
            frac_term = float(np.mean(np.asarray(self.was_terminated, dtype=np.float32))) if len(self.was_terminated) else 0.0

            self.logger.record("custom/ep_len_mean_window", float(lens.mean()))
            self.logger.record("custom/ep_len_p90_window", float(np.percentile(lens, 90)))
            self.logger.record("custom/ep_len_max_window", float(lens.max()))
            self.logger.record("custom/ep_rew_mean_window", float(rews.mean()))
            self.logger.record("custom/ep_rew_p90_window", float(np.percentile(rews, 90)))
            self.logger.record("custom/ep_rew_max_window", float(rews.max()))

            self.logger.record("custom/ep_len_frac_horizon", frac_h)
            self.logger.record("custom/frac_time_limit", frac_tl)
            self.logger.record("custom/frac_terminated", frac_term)

        # -------------------------
        # Actions stats (rollout)
        # -------------------------
        acts = rb.actions
        acts = acts.reshape(-1, acts.shape[-1])  # (n_steps*n_envs, act_dim)
        mean_abs = np.mean(np.abs(acts), axis=0)
        std = np.std(acts, axis=0)
        max_abs = np.max(np.abs(acts), axis=0)

        self.logger.record("debug/actions/mean_abs", float(np.mean(mean_abs)))
        self.logger.record("debug/actions/std", float(np.mean(std)))
        self.logger.record("debug/actions/max_abs", float(np.max(max_abs)))

        if self.log_action_dims:
            for i in range(acts.shape[-1]):
                self.logger.record(f"debug/actions/dim_{i}/mean_abs", float(mean_abs[i]))
                self.logger.record(f"debug/actions/dim_{i}/std", float(std[i]))
                self.logger.record(f"debug/actions/dim_{i}/max_abs", float(max_abs[i]))

        # -------------------------
        # Obs stats (rollout)
        # -------------------------
        obs = rb.observations
        obs = obs.reshape(-1, obs.shape[-1])

        self.logger.record("debug/obs/mean", float(np.mean(obs)))
        self.logger.record("debug/obs/std", float(np.std(obs)))

        # Clip fraction if VecNormalize+clip_obs used (approx: how many features hit +-clip)
        # This is a proxy for “are we saturating normalized obs?”
        clip_obs = getattr(self.model.get_env(), "clip_obs", None)
        if clip_obs is not None:
            clip_frac = float(np.mean(np.abs(obs) >= float(clip_obs) - 1e-6))
            self.logger.record("debug/obs/clip_frac", clip_frac)

        # -------------------------
        # Value / advantage stats
        # -------------------------
        # values: tensor or numpy depending on SB3 version; handle robustly
        values = rb.values
        if isinstance(values, torch.Tensor):
            v = values.detach().cpu().numpy()
        else:
            v = np.asarray(values)
        v = v.reshape(-1)

        adv = rb.advantages
        if isinstance(adv, torch.Tensor):
            a = adv.detach().cpu().numpy()
        else:
            a = np.asarray(adv)
        a = a.reshape(-1)

        self.logger.record("debug/value/vpred_mean", float(v.mean()))
        self.logger.record("debug/value/vpred_std", float(v.std()))
        self.logger.record("debug/adv/mean", float(a.mean()))
        self.logger.record("debug/adv/std", float(a.std()))

        # Returns stats (useful for value scale drift)
        ret = rb.returns
        if isinstance(ret, torch.Tensor):
            rts = ret.detach().cpu().numpy()
        else:
            rts = np.asarray(ret)
        rts = rts.reshape(-1)
        self.logger.record("debug/returns/mean", float(rts.mean()))
        self.logger.record("debug/returns/std", float(rts.std()))

        # Explained variance proxy: 1 - Var(y - yhat) / Var(y)
        # y = returns, yhat = values
        var_y = np.var(rts)
        ev = 0.0 if var_y < 1e-8 else float(1.0 - np.var(rts - v) / (var_y + 1e-8))
        self.logger.record("debug/value/explained_variance_proxy", ev)

        # -------------------------
        # Policy entropy estimate (from rollout obs)
        # -------------------------
        obs_s = obs
        if obs_s.shape[0] > self.entropy_batch:
            idx = np.random.choice(obs_s.shape[0], self.entropy_batch, replace=False)
            obs_s = obs_s[idx]

        obs_t = torch.as_tensor(obs_s, device=self.model.device)
        with torch.no_grad():
            dist = self.model.policy.get_distribution(obs_t)
            ent = dist.distribution.entropy()
            if ent.ndim == 2:
                ent = ent.sum(dim=-1)
            ent_mean = float(ent.mean().item())
        self.logger.record("debug/policy/entropy_mean", ent_mean)

        # -------------------------
        # Optional JSONL mirror
        # -------------------------
        if self._jsonl_f is not None:
            payload = {
                "t": int(self.num_timesteps),
                "time": time.time(),
                "custom/ep_len_mean_window": float(np.mean(self.ep_lens)) if len(self.ep_lens) else None,
                "custom/ep_len_frac_horizon": float(np.mean(np.asarray(self.ep_lens) >= self.horizon)) if len(self.ep_lens) else None,
                "debug/policy/entropy_mean": ent_mean,
                "debug/actions/mean_abs": float(np.mean(mean_abs)),
                "debug/value/explained_variance_proxy": ev,
            }
            self._jsonl_f.write(json.dumps(payload) + "\n")
        self.logger.dump(self.num_timesteps)

    def _on_training_end(self) -> None:
        if self._jsonl_f is not None:
            self._jsonl_f.close()
            self._jsonl_f = None



CKPT_DIR = os.path.join(LOG_DIR, "checkpoints")
callbacks = [
    #LogHopperJointsCallback(),
    CheckpointAndMetadataCallback(),
    #SaveRunMetadataCallback(save_freq=10_000,save_path=CKPT_DIR),
    PPOMetricsCallback(),
]


model.learn(total_timesteps=500_000, callback=callbacks, log_interval=1)
print(f'training Time:{(time.time()-startTime)/3600}')


def rollout_mp4_vecenv(model, vec_env, out_dir, out_name, ep_len=1000, fps=30):

    #os.makedirs(out_dir, exist_ok=True)
    out_path = os.path.join(out_dir, out_name)

    obs = vec_env.reset()
    frames = []
    total = 0.0
    dones = np.zeros(vec_env.num_envs, dtype=bool)

    for _ in range(ep_len):

        # capture frame from VecEnv
        imgs = vec_env.get_images()
        if imgs is not None:
            frames.append(imgs[0])

        action, _ = model.predict(obs, deterministic=True)
        obs, rewards, dones, infos = vec_env.step(action)
        total += float(rewards[0])

        if dones[0]:
            break

    if len(frames) == 0:
        raise RuntimeError("No frames captured from eval_env")

    imageio.mimsave(out_path, frames, fps=fps)

    print(f"Wrote {out_path} return={total:.1f} frames={len(frames)}")

os.makedirs(os.path.join(model.logger.dir, "videos"), exist_ok=True)
rollout_mp4_vecenv(
    model,
    eval_env,
    out_dir=os.path.join(model.logger.dir, "videos"),
    out_name="hopper.mp4"
)


print(f"elapsedTime:{(time.time()-startTime)/60} minutes")
```

```python
import os
import gymnasium as gym
import torch
import time
import imageio
import numpy as np
import json
from collections import deque
from stable_baselines3 import PPO
from stable_baselines3.ppo import MlpPolicy
from stable_baselines3.common.vec_env import SubprocVecEnv, VecMonitor, VecNormalize
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.callbacks import BaseCallback
from stable_baselines3.common.logger import Video


startTime = time.time()

LOG_DIR='/content/hopper'
ENV="Hopper-v5"
NENV_TRAIN = 16
NENV_EVAL= 4

os.makedirs(LOG_DIR, exist_ok=True)

# for colab
os.environ["MUJOCO_GL"] = "egl"          # use EGL (headless GPU)
os.environ["PYOPENGL_PLATFORM"] = "egl"

startTime = time.time()


def make_train_env(env_id, n_envs):
    env = make_vec_env(
        env_id,
        n_envs=n_envs,
        vec_env_cls=SubprocVecEnv,   # ⭐ parallel rollouts
    )
    env = VecNormalize(
        env,
        norm_obs=True,
        norm_reward=False,
        clip_obs=10.0,
    )
    env = VecMonitor(env)
    return env

train_env = make_train_env(env_id=ENV, n_envs=NENV_TRAIN)


def make_eval_env(env_id, train_env, n_envs=NENV_EVAL):
    eval_env = make_vec_env(
        env_id,
        n_envs=n_envs,
        vec_env_cls=SubprocVecEnv,
        env_kwargs={"render_mode": "rgb_array"},
    )
    eval_env = VecNormalize(
        eval_env,
        norm_obs=True,
        norm_reward=False,
    )
    eval_env = VecMonitor(eval_env)

    # ⭐ CRITICAL: freeze normalization and copy stats
    eval_env.obs_rms = train_env.obs_rms
    eval_env.training = False
    eval_env.norm_reward = False
    return eval_env

eval_env = make_eval_env(env_id=ENV, train_env=train_env, n_envs=NENV_EVAL)

policy_kwargs = dict(
    ortho_init=True,
    net_arch=dict(
        activation_fn=torch.nn.Tanh,
        pi=[256, 256],
        vf=[256, 256]
    )
)

model = PPO(
    policy=MlpPolicy,
    env=train_env,
    n_steps=2048,
    batch_size=128,
    n_epochs=5,
    learning_rate=2e-4,
    gamma=0.99,
    gae_lambda=0.95,
    clip_range=0.2,
    target_kl=0.01,
    ent_coef=0.01,
    verbose=1,
    vf_coef=0.5,
    max_grad_norm=0.5,
    policy_kwargs=policy_kwargs,
    device="cuda" if torch.cuda.is_available() else "cpu",
    tensorboard_log=LOG_DIR,
)

print(f"model.policy:{model.policy}")


def rollout_mp4_vecenv(model, vec_env, out_dir, out_name, ep_len=1000, fps=30):
    os.makedirs(out_dir, exist_ok=True)
    out_path = os.path.join(out_dir, out_name)

    obs = vec_env.reset()
    frames = []
    total = 0.0
    dones = np.zeros(vec_env.num_envs, dtype=bool)

    for _ in range(ep_len):
        imgs = vec_env.get_images()
        if imgs is not None:
            frames.append(imgs[0])

        action, _ = model.predict(obs, deterministic=True)
        obs, rewards, dones, infos = vec_env.step(action)
        total += float(rewards[0])

        if dones[0]:
            break

    if len(frames) == 0:
        raise RuntimeError("No frames captured from eval_env")

    imageio.mimsave(out_path, frames, fps=fps)
    print(f"Wrote {out_path} return={total:.1f} frames={len(frames)}")
    return out_path, total, len(frames)


class CheckpointAndMetadataCallback(BaseCallback):
    """
    Saves everything under the current TB run folder:
      <tb_dir>/checkpoints/...
      <tb_dir>/videos/...
      <tb_dir>/artifacts/...
    """
    def __init__(self, eval_env, save_freq=500_000, make_video=True, video_ep_len=1000, video_fps=30, verbose=1):
        super().__init__(verbose)
        self.eval_env = eval_env
        self.save_freq = int(save_freq)
        self.make_video = bool(make_video)
        self.video_ep_len = int(video_ep_len)
        self.video_fps = int(video_fps)

        self.tb_dir = None
        self.ckpt_dir = None
        self.video_dir = None
        self.art_dir = None

    @staticmethod
    def _find_vecnormalize(env):
        """Walk VecEnvWrapper chain to find VecNormalize, or return None."""
        cur = env
        while hasattr(cur, "venv"):
            if isinstance(cur, VecNormalize):
                return cur
            cur = cur.venv
        if isinstance(cur, VecNormalize):
            return cur
        return None

    def _on_training_start(self) -> None:
        # TB run directory (this is the PPO_X folder or tb_log_name_1 folder)
        self.tb_dir = getattr(self.model.logger, "dir", None) or getattr(self.logger, "dir", None)
        if self.tb_dir is None:
            # fallback: keep things together anyway
            self.tb_dir = LOG_DIR

        self.ckpt_dir = os.path.join(self.tb_dir, "checkpoints")
        self.video_dir = os.path.join(self.tb_dir, "videos")
        self.art_dir = os.path.join(self.tb_dir, "artifacts")

        os.makedirs(self.ckpt_dir, exist_ok=True)
        os.makedirs(self.video_dir, exist_ok=True)
        os.makedirs(self.art_dir, exist_ok=True)

        if self.verbose:
            print(f"[run] tb_dir: {self.tb_dir}")
            print(f"[run] ckpt_dir: {self.ckpt_dir}")
            print(f"[run] video_dir: {self.video_dir}")
            print(f"[run] art_dir: {self.art_dir}")

        # Save a run-level config once
        pk = dict(self.model.policy_kwargs)  # shallow copy
        net_arch = pk.get("net_arch", {})
        if isinstance(net_arch, dict) and "activation_fn" in net_arch:
          act = net_arch["activation_fn"]
          if isinstance(act, type):
            net_arch["activation_fn"] = f"{act.__module__}.{act.__name__}"
          else:
            net_arch["activation_fn"] = str(act)
        pk["net_arch"] = net_arch
        run_config = {
            "env_id": ENV,
            "n_env_training": int(NENV_TRAIN),
            "n_env_eval": int(NENV_EVAL),
            "policy_class": self.model.policy.__class__.__name__,
            "policy_kwargs": pk,
            "algo": "PPO",
        }
        with open(os.path.join(self.art_dir, "run_config.json"), "w") as f:
            json.dump(run_config, f, indent=2)

    def _save_bundle(self, t: int) -> None:
        # 1) model
        model_path = os.path.join(self.ckpt_dir, f"model_t{t}.zip")
        self.model.save(model_path)
        if self.verbose:
            print(f"Saved model → {model_path}")

        # 2) VecNormalize
        env = self.model.get_env()
        vecnorm = self._find_vecnormalize(env) if env is not None else None
        if vecnorm is not None:
            vn_path = os.path.join(self.ckpt_dir, f"vecnormalize_t{t}.pkl")
            vecnorm.save(vn_path)
            if self.verbose:
                print(f"Saved VecNormalize → {vn_path}")

        # 3) checkpoint config/metadata
        progress = getattr(self.model, "_current_progress_remaining", 1.0)
        meta = {
            "timesteps": int(t),
            "tb_run_dir": str(self.tb_dir),

            "policy_class": self.model.policy.__class__.__name__,
            "policy_kwargs": self.model.policy_kwargs,

            "n_steps": int(self.model.n_steps),
            "batch_size": int(self.model.batch_size),
            "n_epochs": int(self.model.n_epochs),
            "gamma": float(self.model.gamma),
            "gae_lambda": float(self.model.gae_lambda),

            "learning_rate": float(self.model.lr_schedule(progress)),
            "clip_range": float(
                self.model.clip_range(progress)
                if callable(self.model.clip_range)
                else self.model.clip_range
            ),
            "ent_coef": float(self.model.ent_coef),
            "target_kl": float(self.model.target_kl) if self.model.target_kl is not None else None,
            "vf_coef": float(self.model.vf_coef),
            "max_grad_norm": float(self.model.max_grad_norm),

            "n_env_training": int(NENV_TRAIN),
            "n_env_eval": int(NENV_EVAL),

            "artifact_paths": {
                "model": os.path.join(self.ckpt_dir, f"model_t{t}.zip"),
                "vecnormalize": os.path.join(self.ckpt_dir, f"vecnormalize_t{t}.pkl"),
            }
        }

        json_path = os.path.join(self.ckpt_dir, f"config_t{t}.json")
        with open(json_path, "w") as f:
            json.dump(meta, f, indent=2)
        if self.verbose:
            print(f"[ckpt] saved metadata → {json_path}")

        # 4) optional: eval video tied to checkpoint
        if self.make_video:
            mp4_path, ret, nframes = rollout_mp4_vecenv(
                self.model,
                self.eval_env,
                out_dir=self.video_dir,
                out_name=f"eval_t{t}.mp4",
                ep_len=self.video_ep_len,
                fps=self.video_fps,
            )
            # write a tiny json sidecar so you can index videos quickly
            with open(os.path.join(self.video_dir, f"eval_t{t}.json"), "w") as f:
                json.dump({"timesteps": int(t), "return": float(ret), "frames": int(nframes), "path": mp4_path}, f, indent=2)

    def _on_step(self) -> bool:
        # avoid saving at t=0
        if self.num_timesteps > 0 and (self.num_timesteps % self.save_freq) == 0:
            self._save_bundle(int(self.num_timesteps))
        return True

    def _on_training_end(self) -> None:
        t = int(self.num_timesteps)
        self._save_bundle(t)


class PPOMetricsCallback(BaseCallback):
    def __init__(self, horizon=1000, window=200, entropy_batch=2048, verbose=0):
        super().__init__(verbose)
        self.horizon = horizon
        self.window = window
        self.entropy_batch = entropy_batch
        self.ep_lens = deque(maxlen=window)
        self.ep_rews = deque(maxlen=window)
        self.was_time_limit = deque(maxlen=window)
        self.was_terminated = deque(maxlen=window)

    def _on_step(self) -> bool:
        infos = self.locals.get("infos", [])
        for info in infos:
            ep = info.get("episode", None)
            if ep is None:
                continue
            self.ep_lens.append(float(ep.get("l", 0.0)))
            self.ep_rews.append(float(ep.get("r", 0.0)))
            truncated = bool(info.get("TimeLimit.truncated", False))
            self.was_time_limit.append(truncated)
            self.was_terminated.append(not truncated)
        return True

    def _on_rollout_end(self) -> None:
        rb = self.model.rollout_buffer

        if len(self.ep_lens) > 0:
            lens = np.asarray(self.ep_lens, dtype=np.float32)
            rews = np.asarray(self.ep_rews, dtype=np.float32)

            self.logger.record("custom/ep_len_mean_window", float(lens.mean()))
            self.logger.record("custom/ep_rew_mean_window", float(rews.mean()))
            self.logger.record("custom/ep_len_p90_window", float(np.percentile(lens, 90)))
            self.logger.record("custom/ep_rew_p90_window", float(np.percentile(rews, 90)))

            self.logger.record("custom/ep_len_frac_horizon", float(np.mean(lens >= self.horizon)))
            self.logger.record("custom/frac_time_limit", float(np.mean(np.asarray(self.was_time_limit, dtype=np.float32))) if len(self.was_time_limit) else 0.0)
            self.logger.record("custom/frac_terminated", float(np.mean(np.asarray(self.was_terminated, dtype=np.float32))) if len(self.was_terminated) else 0.0)

        acts = rb.actions.reshape(-1, rb.actions.shape[-1])
        self.logger.record("custom/total_activity", float(np.mean(np.abs(acts))))

        # ensure TB updates even if SB3 logging interval is large
        self.logger.dump(self.num_timesteps)


callbacks = [
    CheckpointAndMetadataCallback(
        eval_env=eval_env,
        save_freq=500_000,
        make_video=True,        # ⭐ videos saved into /content/hopper/PPO_X/videos
        video_ep_len=1000,
        video_fps=30,
        verbose=1,
    ),
    PPOMetricsCallback(verbose=0),
]

model.learn(total_timesteps=500_000, callback=callbacks, log_interval=1)
print(f"training Time:{(time.time()-startTime)/3600}")

# Optional: also save a final human-named video under the current run dir
run_video_dir = os.path.join(model.logger.dir, "videos")
rollout_mp4_vecenv(model, eval_env, out_dir=run_video_dir, out_name="hopper.mp4")

print(f"elapsedTime:{(time.time()-startTime)/60} minutes")
```

```python
import os
import time
import json
import numpy as np
import torch
import imageio
from collections import deque

from stable_baselines3 import PPO
from stable_baselines3.ppo import MlpPolicy
from stable_baselines3.common.vec_env import SubprocVecEnv, VecMonitor, VecNormalize
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.callbacks import BaseCallback


# -----------------------
# Config
# -----------------------
ENV = "Hopper-v5"
NENV_TRAIN = 16
NENV_EVAL = 4

TOTAL_TIMESTEPS = 1_000_000
LEARNING_RATES = [5e-5, 1e-4, 2e-4]   # <- your 3 LRs

SWEEP_ROOT = "/content/hopper_sweeps"  # <- all runs stored here
os.makedirs(SWEEP_ROOT, exist_ok=True)

# Colab mujoco headless
os.environ["MUJOCO_GL"] = "egl"
os.environ["PYOPENGL_PLATFORM"] = "egl"


# -----------------------
# Helpers
# -----------------------
def json_safe(x):
    if x is None or isinstance(x, (str, int, float, bool)):
        return x
    if isinstance(x, dict):
        return {str(k): json_safe(v) for k, v in x.items()}
    if isinstance(x, (list, tuple)):
        return [json_safe(v) for v in x]
    if isinstance(x, type):
        return f"{x.__module__}.{x.__name__}"
    if callable(x):
        return getattr(x, "__qualname__", str(x))
    try:
        import numpy as _np
        if isinstance(x, (_np.integer, _np.floating)):
            return x.item()
    except Exception:
        pass
    return str(x)


def make_train_env(env_id, n_envs):
    env = make_vec_env(env_id, n_envs=n_envs, vec_env_cls=SubprocVecEnv)
    env = VecNormalize(env, norm_obs=True, norm_reward=False, clip_obs=10.0)
    env = VecMonitor(env)
    return env


def make_eval_env(env_id, train_env, n_envs):
    eval_env = make_vec_env(
        env_id,
        n_envs=n_envs,
        vec_env_cls=SubprocVecEnv,
        env_kwargs={"render_mode": "rgb_array"},
    )
    eval_env = VecNormalize(eval_env, norm_obs=True, norm_reward=False)
    eval_env = VecMonitor(eval_env)

    # Freeze normalization, copy stats
    eval_env.obs_rms = train_env.obs_rms
    eval_env.training = False
    eval_env.norm_reward = False
    return eval_env


def rollout_mp4_vecenv(model, vec_env, out_dir, out_name, ep_len=1000, fps=30):
    os.makedirs(out_dir, exist_ok=True)
    out_path = os.path.join(out_dir, out_name)

    obs = vec_env.reset()
    frames = []
    total = 0.0
    dones = np.zeros(vec_env.num_envs, dtype=bool)

    for _ in range(ep_len):
        imgs = vec_env.get_images()
        if imgs is not None:
            frames.append(imgs[0])

        action, _ = model.predict(obs, deterministic=True)
        obs, rewards, dones, infos = vec_env.step(action)
        total += float(rewards[0])

        if dones[0]:
            break

    if frames:
        imageio.mimsave(out_path, frames, fps=fps)

    print(f"[video] Wrote {out_path} return={total:.1f} frames={len(frames)}")
    return out_path, total, len(frames)


# -----------------------
# Callbacks (minimal, but stores into TB run dir)
# -----------------------
class CheckpointAndMetadataCallback(BaseCallback):
    """
    Saves under current TB run directory:
      <tb_dir>/checkpoints/...
      <tb_dir>/videos/...
      <tb_dir>/artifacts/...
    """
    def __init__(self, eval_env, save_freq=500_000, make_video=True, verbose=1):
        super().__init__(verbose)
        self.eval_env = eval_env
        self.save_freq = int(save_freq)
        self.make_video = bool(make_video)

        self.tb_dir = None
        self.ckpt_dir = None
        self.video_dir = None
        self.art_dir = None

    @staticmethod
    def _find_vecnormalize(env):
        cur = env
        while hasattr(cur, "venv"):
            if isinstance(cur, VecNormalize):
                return cur
            cur = cur.venv
        return cur if isinstance(cur, VecNormalize) else None

    def _on_training_start(self) -> None:
        self.tb_dir = getattr(self.model.logger, "dir", None) or getattr(self.logger, "dir", None)
        if self.tb_dir is None:
            raise RuntimeError("Could not determine TB directory (logger.dir is None)")

        self.ckpt_dir = os.path.join(self.tb_dir, "checkpoints")
        self.video_dir = os.path.join(self.tb_dir, "videos")
        self.art_dir = os.path.join(self.tb_dir, "artifacts")
        os.makedirs(self.ckpt_dir, exist_ok=True)
        os.makedirs(self.video_dir, exist_ok=True)
        os.makedirs(self.art_dir, exist_ok=True)

        if self.verbose:
            print(f"[run] tb_dir: {self.tb_dir}")
            print(f"[run] ckpt_dir: {self.ckpt_dir}")
            print(f"[run] video_dir: {self.video_dir}")
            print(f"[run] art_dir: {self.art_dir}")

        # Save run config once
        run_config = {
            "env_id": ENV,
            "n_env_training": int(NENV_TRAIN),
            "n_env_eval": int(NENV_EVAL),
            "policy_class": self.model.policy.__class__.__name__,
            "policy_kwargs": json_safe(self.model.policy_kwargs),
        }
        with open(os.path.join(self.art_dir, "run_config.json"), "w") as f:
            json.dump(run_config, f, indent=2)

    def _save_bundle(self, t: int) -> None:
        # model
        model_path = os.path.join(self.ckpt_dir, f"model_t{t}.zip")
        self.model.save(model_path)

        # vecnormalize
        env = self.model.get_env()
        vecnorm = self._find_vecnormalize(env) if env is not None else None
        vn_path = None
        if vecnorm is not None:
            vn_path = os.path.join(self.ckpt_dir, f"vecnormalize_t{t}.pkl")
            vecnorm.save(vn_path)

        # metadata
        progress = getattr(self.model, "_current_progress_remaining", 1.0)
        meta = {
            "timesteps": int(t),
            "tb_run_dir": str(self.tb_dir),
            "learning_rate_now": float(self.model.lr_schedule(progress)),
            "clip_range_now": float(self.model.clip_range(progress)) if callable(self.model.clip_range) else float(self.model.clip_range),
            "hyperparams": {
                "n_steps": int(self.model.n_steps),
                "batch_size": int(self.model.batch_size),
                "n_epochs": int(self.model.n_epochs),
                "gamma": float(self.model.gamma),
                "gae_lambda": float(self.model.gae_lambda),
                "ent_coef": float(self.model.ent_coef),
                "vf_coef": float(self.model.vf_coef),
                "max_grad_norm": float(self.model.max_grad_norm),
                "target_kl": float(self.model.target_kl) if self.model.target_kl is not None else None,
            },
            "paths": {"model": model_path, "vecnormalize": vn_path},
        }
        with open(os.path.join(self.ckpt_dir, f"config_t{t}.json"), "w") as f:
            json.dump(json_safe(meta), f, indent=2)

        if self.make_video:
            rollout_mp4_vecenv(
                self.model, self.eval_env,
                out_dir=self.video_dir,
                out_name=f"eval_t{t}.mp4",
                ep_len=1000, fps=30
            )

        if self.verbose:
            print(f"[ckpt] saved bundle at t={t}")

    def _on_step(self) -> bool:
        if self.num_timesteps > 0 and (self.num_timesteps % self.save_freq) == 0:
            self._save_bundle(int(self.num_timesteps))
        return True

    def _on_training_end(self) -> None:
        self._save_bundle(int(self.num_timesteps))


# -----------------------
# Sweep loop
# -----------------------
def run_one_lr(lr: float):
    # Each LR gets its own directory root
    run_root = os.path.join(SWEEP_ROOT, f"lr_{lr:g}")
    os.makedirs(run_root, exist_ok=True)

    # Use tensorboard_log=run_root, and tb_log_name fixed -> produces run_root/run/
    # (avoids PPO_1/2/... clutter across runs)
    tb_log_name = "run"

    train_env = make_train_env(ENV, NENV_TRAIN)
    eval_env = make_eval_env(ENV, train_env, NENV_EVAL)

    policy_kwargs = dict(
        ortho_init=True,
        net_arch=dict(
            activation_fn=torch.nn.Tanh,
            pi=[256, 256],
            vf=[256, 256],
        )
    )

    model = PPO(
        policy=MlpPolicy,
        env=train_env,
        n_steps=2048,
        batch_size=128,
        n_epochs=5,
        learning_rate=lr,
        gamma=0.99,
        gae_lambda=0.95,
        clip_range=0.2,
        target_kl=0.03,
        ent_coef=0.02,
        vf_coef=0.5,
        max_grad_norm=0.5,
        policy_kwargs=policy_kwargs,
        device="cuda" if torch.cuda.is_available() else "cpu",
        tensorboard_log=run_root,   # <- separate per LR
        verbose=0,
    )

    callbacks = [
        CheckpointAndMetadataCallback(eval_env=eval_env, save_freq=200_000, make_video=True, verbose=1),
    ]

    t0 = time.time()
    model.learn(
        total_timesteps=TOTAL_TIMESTEPS,
        callback=callbacks,
        log_interval=1,
        tb_log_name=tb_log_name,         # <- keep run folder name stable
        reset_num_timesteps=True,        # <- each LR run starts at 0
    )

    # Save an extra named final video
    video_dir = os.path.join(model.logger.dir, "videos")
    rollout_mp4_vecenv(model, eval_env, out_dir=video_dir, out_name="final.mp4")

    # Write a simple summary file at lr root
    summary = {
        "learning_rate": lr,
        "total_timesteps": TOTAL_TIMESTEPS,
        "tb_dir": model.logger.dir,
        "elapsed_minutes": (time.time() - t0) / 60.0,
    }
    with open(os.path.join(run_root, "summary.json"), "w") as f:
        json.dump(summary, f, indent=2)

    # cleanup
    train_env.close()
    eval_env.close()

    return summary


all_summaries = []
for lr in LEARNING_RATES:
    print("\n" + "="*80)
    print(f"SWEEP lr={lr}")
    print("="*80)
    all_summaries.append(run_one_lr(lr))

print("\nDONE. Summaries:")
for s in all_summaries:
    print(s)
```

```python
import os
import gymnasium as gym
import torch
import numpy as np
from stable_baselines3 import PPO
from stable_baselines3.ppo import MlpPolicy
from stable_baselines3.common.vec_env import SubprocVecEnv, VecMonitor, VecNormalize
from stable_baselines3.common.env_util import make_vec_env

# Change these:
from vizier import pyvizier as core_vz
from vizier.service import clients
from vizier.service import pyvizier as vz  # Use this for EVERYTHING (config + service)

# --- Global Config ---
LOG_DIR = '/content/hopper_vizier'
ENV_ID = "Hopper-v5"
TRAIN_STEPS = 300_000  # Reduced for faster optimization cycles

os.makedirs(LOG_DIR, exist_ok=True)
os.environ["MUJOCO_GL"] = "egl"

# --- Helper Functions (Same as your original) ---
def make_train_env(env_id, n_envs):
    env = make_vec_env(env_id, n_envs=n_envs, vec_env_cls=SubprocVecEnv)
    env = VecNormalize(env, norm_obs=True, norm_reward=False, clip_obs=10.0)
    return VecMonitor(env)

# --- The Objective Function ---
def train_ppo_trial(learning_rate: float, ent_coef: float, batch_size: int, n_steps: int) -> float:
    """
    This function represents one 'Trial' in Vizier.
    It returns the performance metric Vizier wants to maximize.
    """
    train_env = make_train_env(ENV_ID, n_envs=16)

    policy_kwargs = dict(
        ortho_init=True,
        net_arch=dict(activation_fn=torch.nn.Tanh, pi=[256, 256], vf=[256, 256])
    )

    model = PPO(
        policy=MlpPolicy,
        env=train_env,
        learning_rate=learning_rate,
        ent_coef=ent_coef,
        batch_size=batch_size,
        n_steps=n_steps,
        n_epochs=10,
        gamma=0.99,
        policy_kwargs=policy_kwargs,
        device="cuda",
        verbose=0 # Keep logs clean during optimization
    )

    model.learn(total_timesteps=TRAIN_STEPS)

    # Calculate Metric: Mean reward of the last 100 episodes
    # SB3's VecMonitor stores episode rewards in the 'get_episode_rewards()' list
    ep_rewards = train_env.get_episode_rewards()
    final_performance = np.mean(ep_rewards[-50:]) if len(ep_rewards) >= 50 else -1000.0

    train_env.close()
    return float(final_performance)

# --- Vizier Setup ---
def main():
    # 1. StudyConfig and Search Space
    study_config = vz.StudyConfig(algorithm='DEFAULT')
    root = study_config.search_space.root

    # ⭐️ FIX: Use 'scale_type' and 'vz.ScaleType'
    root.add_float_param(
        'learning_rate',
        1e-5, 1e-3,
        scale_type=vz.ScaleType.LOG  # Changed from 'scaling_type'
    )

    root.add_float_param('ent_coef', 0.0, 0.05)
    root.add_discrete_param('batch_size', [64, 128, 256])
    root.add_discrete_param('n_steps', [1024, 2048])

    # 2. Metric Setup
    study_config.metric_information.append(
        vz.MetricInformation(
            'mean_reward',
            goal=vz.ObjectiveMetricGoal.MAXIMIZE
        )
    )

    # 3. Setup Study Client
    study = clients.Study.from_study_config(
        study_config,
        owner='hopper_expert',
        study_id='ppo_hopper_vfinal'
    )
    # 3. Optimization Loop
    for i in range(10):  # Number of iterations
        print(f"iteration:{i}")
        suggestions = study.suggest(count=1)
        for suggestion in suggestions:
            print(f"\n--- Starting Trial {i+1} ---")
            params = suggestion.parameters

            # Execute Training
            reward = train_ppo_trial(
                learning_rate=params['learning_rate'],
                ent_coef=params['ent_coef'],
                batch_size=int(params['batch_size']),
                n_steps=int(params['n_steps'])
            )

            # Report back to Vizier
            suggestion.complete(vz.Measurement({'mean_reward': reward}))
            print(f"Trial {i+1} Result: {reward:.2f} (LR: {params['learning_rate']:.5f})")

    # 4. Results
    best_trial = list(study.optimal_trials())[0]
    print(f"\nOptimization Finished!")
    print(f"Best Reward: {best_trial.final_measurement.metrics['mean_reward'].value}")
    print(f"Best Params: {best_trial.parameters}")

if __name__ == "__main__":
    main()
```
