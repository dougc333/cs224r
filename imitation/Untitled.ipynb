```python
"""This is a simple example demonstrating how to clone the behavior of an expert.

Refer to the jupyter notebooks for more detailed examples of how to use the algorithms.
"""
import numpy as np
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.ppo import MlpPolicy

from imitation.algorithms import bc
from imitation.data import rollout
from imitation.data.wrappers import RolloutInfoWrapper
from imitation.policies.serialize import load_policy
from imitation.util.util import make_vec_env

from huggingface_hub import hf_hub_download
from stable_baselines3 import PPO, SAC

import gymnasium as gym
import imageio.v2 as imageio
import numpy as np


rng = np.random.default_rng(0)
ENV_ID = "Ant-v4"

env = make_vec_env(
    ENV_ID,
    rng=rng,
    n_envs=1,
    post_wrappers=[lambda env, _: RolloutInfoWrapper(env)],  # for computing rollouts
)


def train_expert():
    # note: use `download_expert` instead to download a pretrained, competent expert
    print("Training a expert.")
    expert = PPO(
        policy=MlpPolicy,
        env=env,
        seed=0,
        batch_size=64,
        ent_coef=0.0,
        learning_rate=0.0003,
        n_epochs=10,
        n_steps=64,
    )
    expert.learn(1_000)  # Note: change this to 100_000 to train a decent expert.
    return expert

def download_expert():
    print("Downloading SB3 expert zip...")
    zip_path = hf_hub_download(
        repo_id="Mzou000/PPO-Ant-v4",
        filename="Ant-v4.zip",
    )

    try:
        expert = PPO.load(zip_path, env=env)
        print("Loaded as PPO")
        return expert
    except Exception as e:
        print("PPO.load failed -> trying SAC. Error:", repr(e))
        expert = SAC.load(zip_path, env=env)
        print("Loaded as SAC")
        return expert


def sample_expert_transitions():
    # expert = train_expert()  # uncomment to train your own expert
    expert = download_expert()

    print("Sampling expert transitions.")
    rollouts = rollout.rollout(
        expert,
        env,
        rollout.make_sample_until(min_timesteps=None, min_episodes=50),
        rng=rng,
    )
    return rollout.flatten_trajectories(rollouts)


transitions = sample_expert_transitions()
bc_trainer = bc.BC(
    observation_space=env.observation_space,
    action_space=env.action_space,
    demonstrations=transitions,
    rng=rng,
)

evaluation_env = make_vec_env(
    ENV_ID,
    rng=rng,
    n_envs=1,
    #env_make_kwargs={"render_mode": "human"},  # for rendering
)

print("Evaluating the untrained policy.")
reward, _ = evaluate_policy(
    bc_trainer.policy,  # type: ignore[arg-type]
    evaluation_env,
    n_eval_episodes=3,
    render=False,  # comment out to speed up
)
print(f"Reward before training: {reward}")

print("Training a policy using Behavior Cloning")
bc_trainer.train(n_epochs=1)

print("Evaluating the trained policy.")
reward, _ = evaluate_policy(
    bc_trainer.policy,  # type: ignore[arg-type]
    evaluation_env,
    n_eval_episodes=3,
    render=True,  # comment out to speed up
)
print(f"Reward after training: {reward}")
```

```python
import numpy as np
from huggingface_hub import hf_hub_download
from stable_baselines3 import PPO, SAC
from stable_baselines3.common.evaluation import evaluate_policy
from stable_baselines3.ppo import MlpPolicy

from imitation.algorithms import bc
from imitation.data import rollout
from imitation.data.wrappers import RolloutInfoWrapper
from imitation.util.util import make_vec_env

import gymnasium as gym
import imageio.v2 as imageio

rng = np.random.default_rng(0)
ENV_ID = "Ant-v4"

# Data collection / training env (VecEnv)
env = make_vec_env(
    ENV_ID,
    rng=rng,
    n_envs=1,
    post_wrappers=[lambda e, _: RolloutInfoWrapper(e)],
)

def download_expert():
    print("Downloading SB3 expert zip...")
    zip_path = hf_hub_download("Mzou000/PPO-Ant-v4", filename="Ant-v4.zip")
    try:
        expert = PPO.load(zip_path, env=env)
        print("Loaded as PPO")
        return expert
    except Exception as e:
        print("PPO.load failed -> trying SAC. Error:", repr(e))
        expert = SAC.load(zip_path, env=env)
        print("Loaded as SAC")
        return expert

def sample_expert_transitions():
    expert = download_expert()
    print("Sampling expert transitions.")
    rollouts_list = rollout.rollout(
        expert,
        env,
        rollout.make_sample_until(min_timesteps=None, min_episodes=50),
        rng=rng,
    )
    expert_reward, _ = evaluate_policy(expert, evaluation_env, n_eval_episodes=10, render=False)
    print("Expert avg return:", expert_reward)
    return rollout.flatten_trajectories(rollouts_list)

transitions = sample_expert_transitions()

bc_trainer = bc.BC(
    observation_space=env.observation_space,
    action_space=env.action_space,
    demonstrations=transitions,
    rng=rng,
)

# Numeric evaluation (VecEnv is fine here)
evaluation_env = make_vec_env(ENV_ID, rng=rng, n_envs=1)

print("Evaluating the untrained policy (no rendering).")
reward, _ = evaluate_policy(bc_trainer.policy, evaluation_env, n_eval_episodes=10, render=False)
print("Reward before training:", reward)

print("Training BC")
bc_trainer.train(n_epochs=20)

print("Evaluating the trained policy (no rendering).")
reward, _ = evaluate_policy(bc_trainer.policy, evaluation_env, n_eval_episodes=10, render=False)
print("Reward after training:", reward)

# MP4 rendering env (plain Gym env)
mp4_env = gym.make(ENV_ID, render_mode="rgb_array")

obs, info = mp4_env.reset()
frame = mp4_env.render()
print("mp4 env frame shape:", np.asarray(frame).shape)

def rollout_frames_sb3(policy, env, ep_len=1000):
    obs, info = env.reset()
    frames = []
    total = 0.0
    for _ in range(ep_len):
        frame = env.render()
        if frame is not None:
            frames.append(frame)
        action, _ = policy.predict(obs, deterministic=True)
        obs, reward, terminated, truncated, info = env.step(action)
        total += float(reward)
        if terminated or truncated:
            break
    return total, frames

ret, frames = rollout_frames_sb3(bc_trainer.policy, mp4_env)
print("MP4 rollout return:", ret)

with imageio.get_writer("bc_ant.mp4", fps=30, codec="libx264", quality=8) as w:
    for f in frames:
        w.append_data(np.asarray(f, dtype=np.uint8))

print("Wrote bc_ant.mp4")
dir(mp4_env.unwrapped)[:50]
mp4_env.close()
evaluation_env.close()
```

```python

import gymnasium as gym
from stable_baselines3 import PPO
from stable_baselines3.ppo import MlpPolicy

env = gym.make("Ant-v4")

model = PPO(
    policy=MlpPolicy,
    env=env,
    n_steps=2048,          # important for Ant
    batch_size=64,
    n_epochs=10,
    learning_rate=3e-4,
    gamma=0.99,
    gae_lambda=0.95,
    clip_range=0.2,
    ent_coef=0.0,
    verbose=1,
)

model.learn(total_timesteps=500_000)




def rollout_mp4(policy, env_id="Ant-v4", filename="ant.mp4", ep_len=1000):
    env = gym.make(env_id, render_mode="rgb_array")
    obs, info = env.reset()

    frames = []
    total_reward = 0.0

    for _ in range(ep_len):
        frame = env.render()
        if frame is not None:
            frames.append(frame)

        action, _ = policy.predict(obs, deterministic=True)
        obs, reward, terminated, truncated, info = env.step(action)
        total_reward += float(reward)

        if terminated or truncated:
            break

    env.close()

    with imageio.get_writer(filename, fps=30, codec="libx264") as w:
        for f in frames:
            w.append_data(np.asarray(f, dtype=np.uint8))

    print(f"Wrote {filename}, return={total_reward:.1f}")
    return total_reward


rollout_mp4(model, filename="ppo_ant.mp4")

```
